```python
import pandas as pd
```

```python
df2021 = pd.read_csv("2021.csv")
df2022 = pd.read_csv('2022.csv')
df2023 = pd.read_csv('2023.csv')
df2024 = pd.read_csv('2024.csv')
df2025 = pd.read_csv('2025.csv')
```

```python
df = pd.concat(
    [df2021, df2022, df2023, df2024, df2025],
    ignore_index=True
)

print("Total rows and columns:", df.shape)
print("\nColumn names:")
print(df.columns.tolist())

df.head()
```

```python
haryana_df = df[df["State"].astype(str).str.strip().str.lower() == "haryana"].copy()

print("Haryana rows:", haryana_df.shape)

print("\nHaryana date range:")
print("From:", haryana_df["Arrival_Date"].min())
print("To:", haryana_df["Arrival_Date"].max())

print("\nTotal unique commodities:", haryana_df["Commodity"].nunique())

print("\nCommodities available in Haryana:")
print(sorted(haryana_df["Commodity"].dropna().unique()))
```

```python
keywords = ["wheat", "paddy", "rice", "bajra", "mustard", "cotton"]

for word in keywords:
    matches = haryana_df[
        haryana_df["Commodity"]
        .str.contains(word, case=False, na=False)
    ]["Commodity"].unique()

    print(f"\n{word.upper()}:")
    print(matches)
```

```python
selected_commodities = [
    "Wheat",
    "Paddy (Dhan)(Basmati)",
    "Paddy (Dhan)(Common)",
    "Bajra (Pearl Millet/Cumbu)",
    "Mustard",
    "Cotton"
]

crop_df = haryana_df[
    haryana_df["Commodity"].isin(selected_commodities)
].copy()

print("Total selected rows:", crop_df.shape)

print("\nRows per commodity:")
print(crop_df["Commodity"].value_counts())
```

```python
crop_df["Arrival_Date"] = pd.to_datetime(
    crop_df["Arrival_Date"],
    errors="coerce"
)

crop_df["Year"] = crop_df["Arrival_Date"].dt.year

year_crop_count = pd.crosstab(
    crop_df["Year"],
    crop_df["Commodity"]
)

print(year_crop_count)
```

```python
print("DATA TYPES:")
print(crop_df.dtypes)

print("\nMISSING VALUES:")
print(crop_df.isnull().sum())

print("\nDUPLICATE ROWS:")
print(crop_df.duplicated().sum())

print("\nDATASET SIZE:")
print(crop_df.shape)
```

```python
print("Duplicate rows:", crop_df.duplicated().sum())

print("\nZero or negative prices:")
print("Min Price:", (crop_df["Min_Price"] <= 0).sum())
print("Max Price:", (crop_df["Max_Price"] <= 0).sum())
print("Modal Price:", (crop_df["Modal_Price"] <= 0).sum())

print("\nInvalid price relationships:")

print(
    "Min Price > Max Price:",
    (crop_df["Min_Price"] > crop_df["Max_Price"]).sum()
)

print(
    "Modal Price < Min Price:",
    (crop_df["Modal_Price"] < crop_df["Min_Price"]).sum()
)

print(
    "Modal Price > Max Price:",
    (crop_df["Modal_Price"] > crop_df["Max_Price"]).sum()
)
```

```python
problem_rows = crop_df[
    (crop_df["Min_Price"] <= 0) |
    (crop_df["Max_Price"] <= 0) |
    (crop_df["Modal_Price"] <= 0) |
    (crop_df["Min_Price"] > crop_df["Max_Price"]) |
    (crop_df["Modal_Price"] < crop_df["Min_Price"]) |
    (crop_df["Modal_Price"] > crop_df["Max_Price"])
]

print("Total problematic rows:", len(problem_rows))

problem_rows[
    [
        "Arrival_Date",
        "District",
        "Market",
        "Commodity",
        "Variety",
        "Min_Price",
        "Max_Price",
        "Modal_Price"
    ]
]
```

```python
clean_df = crop_df[
    (crop_df["Min_Price"] > 0) &
    (crop_df["Max_Price"] > 0) &
    (crop_df["Modal_Price"] > 0) &
    (crop_df["Min_Price"] <= crop_df["Modal_Price"]) &
    (crop_df["Modal_Price"] <= crop_df["Max_Price"])
].copy()

print("Before cleaning:", crop_df.shape)
print("After cleaning:", clean_df.shape)
print("Rows removed:", len(crop_df) - len(clean_df))
```

```python
clean_df.to_csv("haryana_crop_prices_cleaned.csv", index=False)
```

```python
wheat_df = clean_df[
    clean_df["Commodity"] == "Wheat"
].copy()

bajra_df = clean_df[
    clean_df["Commodity"] == "Bajra (Pearl Millet/Cumbu)"
].copy()

mustard_df = clean_df[
    clean_df["Commodity"] == "Mustard"
].copy()

cotton_df = clean_df[
    clean_df["Commodity"] == "Cotton"
].copy()

paddy_basmati_df = clean_df[
    clean_df["Commodity"] == "Paddy (Dhan)(Basmati)"
].copy()

paddy_common_df = clean_df[
    clean_df["Commodity"] == "Paddy (Dhan)(Common)"
].copy()
```
